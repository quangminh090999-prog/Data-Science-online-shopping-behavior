```python
import pandas as pd
from pathlib import Path

# Tìm file dữ liệu
file_path = Path("../data/processed/online_retail_sales.csv")

if not file_path.exists():
    file_path = Path("data/processed/online_retail_sales.csv")

# Đọc dữ liệu
df = pd.read_csv(file_path)

# Kiểm tra dữ liệu
print("Kích thước dữ liệu:", df.shape)

print("\nCác cột hiện có:")
print(df.columns.tolist())

print("\nKiểu dữ liệu:")
print(df.dtypes)

print("\n5 dòng đầu tiên:")
display(df.head())
```

```python
# Chuẩn hóa mã hóa đơn
df["Invoice"] = df["Invoice"].astype("string")

# Chuyển InvoiceDate sang kiểu datetime
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

# Kiểm tra dữ liệu sau chuyển đổi
print("Kiểu dữ liệu sau chuẩn hóa:")
print(df[["Invoice", "InvoiceDate"]].dtypes)

print("\nSố giá trị InvoiceDate không hợp lệ:",
      df["InvoiceDate"].isna().sum())
```

```python
# Tạo đặc trưng thời gian
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Day"] = df["InvoiceDate"].dt.day
df["DayOfWeek"] = df["InvoiceDate"].dt.dayofweek
df["Hour"] = df["InvoiceDate"].dt.hour

# Kiểm tra kết quả
display(
    df[[
        "InvoiceDate",
        "Year",
        "Month",
        "Day",
        "DayOfWeek",
        "Hour"
    ]].head(10)
)
```

```python
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Day"] = df["InvoiceDate"].dt.day
df["DayOfWeek"] = df["InvoiceDate"].dt.dayofweek
df["Hour"] = df["InvoiceDate"].dt.hour

display(df[[
    "InvoiceDate",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "Hour"
]].head(10))
```

```python
# Tổng số đơn hàng của mỗi khách hàng
df["TotalOrders"] = (
    df.groupby("Customer ID")["Invoice"]
    .transform("nunique")
)

# Tổng doanh thu của mỗi khách hàng
df["CustomerRevenue"] = (
    df.groupby("Customer ID")["Revenue"]
    .transform("sum")
)
```

```python
# Hiển thị thông tin của một số khách hàng
display(
    df[[
        "Customer ID",
        "Invoice",
        "Revenue",
        "TotalOrders",
        "CustomerRevenue"
    ]].head(15)
)
```

```python
print("Số khách hàng:", df["Customer ID"].nunique())

print("\nSố giá trị TotalOrders bị thiếu:",
      df["TotalOrders"].isna().sum())

print("Số giá trị CustomerRevenue bị thiếu:",
      df["CustomerRevenue"].isna().sum())

print("\nThống kê TotalOrders:")
print(df["TotalOrders"].describe())

print("\nThống kê CustomerRevenue:")
print(df["CustomerRevenue"].describe())
```

```python
# Chỉ lấy các giao dịch có Customer ID
customer_df = df.dropna(subset=["Customer ID"]).copy()

# Ngày tham chiếu: một ngày sau giao dịch cuối cùng
snapshot_date = customer_df["InvoiceDate"].max() + pd.Timedelta(days=1)

# Tính các chỉ số RFM
rfm = customer_df.groupby("Customer ID").agg(
    Recency=("InvoiceDate", lambda x: (snapshot_date - x.max()).days),
    Frequency=("Invoice", "nunique"),
    Monetary=("Revenue", "sum")
).reset_index()

# Hiển thị kết quả
display(rfm.head(10))
```

```python
rfm.to_csv(
    "../data/processed/customer_rfm.csv",
    index=False,
    encoding="utf-8-sig"
)
```
