```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Đường dẫn dữ liệu
file_path = "../data/online_retail_II.xlsx"

# Kiểm tra các sheet trong file Excel
excel_file = pd.ExcelFile(file_path)

print("Danh sách sheet:")
print(excel_file.sheet_names)

# Kiểm tra kích thước từng sheet
for sheet in excel_file.sheet_names:
    temp = pd.read_excel(file_path, sheet_name=sheet)

    print(f"\nSheet: {sheet}")
    print("Kích thước:", temp.shape)
    print("Các cột:", temp.columns.tolist())
```

```python

# Đọc dữ liệu từ hai sheet
df_2009 = pd.read_excel(
    file_path,
    sheet_name="Year 2009-2010"
)

df_2010 = pd.read_excel(
    file_path,
    sheet_name="Year 2010-2011"
)

# Đánh dấu nguồn dữ liệu
df_2009["SourceSheet"] = "Year 2009-2010"
df_2010["SourceSheet"] = "Year 2010-2011"

# Gộp hai sheet
df = pd.concat(
    [df_2009, df_2010],
    ignore_index=True
)

# Kiểm tra kết quả
print("Kích thước dữ liệu sau khi gộp:", df.shape)

display(df.head())
display(df.tail())
```

```python

# 1. Kiểm tra thông tin dữ liệu
print("===== DATA INFORMATION =====")
df.info()

# 2. Thống kê mô tả
print("\n===== DESCRIPTIVE STATISTICS =====")
display(df.describe(include="all").T)

# 3. Kiểm tra giá trị thiếu
print("\n===== MISSING VALUES =====")
missing = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing Values": missing,
    "Missing Percentage (%)": (
        missing / len(df) * 100
    ).round(2)
})

display(missing_summary)

# 4. Kiểm tra dữ liệu trùng lặp
print("\n===== DUPLICATE VALUES =====")
print("Số dòng trùng lặp:", df.duplicated().sum())

# 5. Kiểm tra kiểu dữ liệu
print("\n===== DATA TYPES =====")
display(df.dtypes)
```

```python

# 1. Khoảng thời gian dữ liệu
print("===== DATE RANGE =====")
print("Ngày bắt đầu:", df["InvoiceDate"].min())
print("Ngày kết thúc:", df["InvoiceDate"].max())

# 2. Số lượng đối tượng khác nhau
print("\n===== UNIQUE VALUES =====")
print("Số hóa đơn:", df["Invoice"].nunique())
print("Số sản phẩm:", df["StockCode"].nunique())
print("Số khách hàng:", df["Customer ID"].nunique())
print("Số quốc gia:", df["Country"].nunique())

# 3. Kiểm tra hóa đơn hủy
cancelled = df[
    df["Invoice"].astype("string").str.startswith("C", na=False)
]

print("\n===== CANCELLED INVOICES =====")
print("Số dòng thuộc hóa đơn hủy:", len(cancelled))
print("Số hóa đơn hủy:", cancelled["Invoice"].nunique())

# 4. Kiểm tra Quantity
print("\n===== QUANTITY ANALYSIS =====")
print("Quantity âm:", (df["Quantity"] < 0).sum())
print("Quantity bằng 0:", (df["Quantity"] == 0).sum())
print("Quantity dương:", (df["Quantity"] > 0).sum())

# 5. Kiểm tra Price
print("\n===== PRICE ANALYSIS =====")
print("Price âm:", (df["Price"] < 0).sum())
print("Price bằng 0:", (df["Price"] == 0).sum())
print("Price dương:", (df["Price"] > 0).sum())
```

```python

# Tạo điều kiện nhận diện hóa đơn hủy
is_cancelled = (
    df["Invoice"].astype("string").str.startswith("C", na=False)
)

# Tạo điều kiện Quantity âm
is_return = df["Quantity"] < 0

# Bảng phân loại giao dịch
transaction_analysis = pd.crosstab(
    is_cancelled,
    is_return,
    rownames=["Cancelled Invoice"],
    colnames=["Negative Quantity"]
)

display(transaction_analysis)
```

```python
# Lấy các dòng Quantity âm nhưng không thuộc hóa đơn hủy
negative_not_cancelled = df[
    (df["Quantity"] < 0)
    & (~df["Invoice"].astype(str).str.startswith("C"))
]

print("Số dòng:", len(negative_not_cancelled))

# Xem 20 dòng đầu
display(negative_not_cancelled.head(20))

# Thống kê Description xuất hiện nhiều nhất
display(
    negative_not_cancelled["Description"]
    .value_counts(dropna=False)
    .head(20)
)
```

```python
# Lấy các dòng trùng lặp hoàn toàn
duplicate_rows = df[
    df.duplicated(keep=False)
].copy()

# Thống kê số lần xuất hiện của từng nhóm trùng
duplicate_groups = (
    duplicate_rows
    .groupby(list(df.columns), dropna=False)
    .size()
    .reset_index(name="Count")
)

print("Số nhóm dữ liệu trùng:", len(duplicate_groups))

print("Số dòng thuộc nhóm trùng:", len(duplicate_rows))

print("Số dòng dư thừa nếu giữ lại một bản ghi mỗi nhóm:",
      (duplicate_groups["Count"] - 1).sum())

display(duplicate_groups.head(20))
```

```python
# Thống kê số nhóm theo số lần xuất hiện
display(
    duplicate_groups["Count"]
    .value_counts()
    .sort_index()
)
```

```python
# Tìm bản ghi xuất hiện 20 lần
display(
    duplicate_groups[
        duplicate_groups["Count"] == 20
    ]
)
```
